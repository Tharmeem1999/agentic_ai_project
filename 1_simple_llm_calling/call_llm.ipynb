```python
from dotenv import load_dotenv
from langchain_groq import ChatGroq

load_dotenv()
```

```python
llm = ChatGroq(model = "llama-3.3-70b-versatile")

response = llm. invoke("How many moons saturn have?")
print(response.text)
```

```python
response = llm. invoke([
    ["system", "You are a helpful assistant that answers in short one line"],
    ["human", "How many moons saturn have?"]
])
print(response.text)
```

## Temperature

```python
llm = ChatGroq(model = "llama-3.3-70b-versatile", temperature=0)

prompt = "Write a single tagline for a quirky coffee shop. Only output the tagline, nothing else"

response = llm.invoke(prompt)
print(response.text)
```

```python
llm = ChatGroq(model = "llama-3.3-70b-versatile", temperature=1)

prompt = "Write a single tagline for a quirky coffee shop. Only output the tagline, nothing else"

response = llm.invoke(prompt)
print(response.text)
```
